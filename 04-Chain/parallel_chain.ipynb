### What is Parallel Chain in LangChain?

Parallel Chain is a way to execute multiple chains in parallel using LangChain. It uses the `concurrent.futures` library to run multiple chains in separate threads, allowing for faster execution of multiple chains.

Parallel Chain is useful when you have multiple chains that do not depend on each other, and you want to execute them all at once. For example, you could use Parallel Chain to execute multiple retrieval chains in parallel, or to execute multiple generation chains in parallel.



```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_groq import ChatGroq
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnableParallel
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
model1=ChatGoogleGenerativeAI(model="gemini-2.5-flash-lite", temperature=0.7, max_tokens=2048)
model2=ChatGroq(model="llama-3.1-8b-instant", temperature=0.7, max_tokens=2048)
```

```python
prompt1=PromptTemplate(
    input_variables=["text"],
    template="Explan the following text in teachable way:\n\n{text}\n\nExplanation:")
prompt2=PromptTemplate(
    input_variables=["text"],
    template="What are the key points in the following text:\n\n{text}\n\nKey Points:")

```

```python
parser=StrOutputParser()
```

```python
chain1=prompt1 | model1 | parser
chain2=prompt2 | model2 | parser
```

```python
parallel_chain=RunnableParallel({"chain1": chain1, "chain2": chain2})
```

```python
parallel_chain.get_graph().print_ascii()
```

```python
response=parallel_chain.invoke({"text": "LangChain is a framework for developing applications powered by language models. It provides a standard interface for all components and a powerful standard library of components. It can be used for chatbots, Generative Question-Answering (GQA), summarization, and much more."})
```
