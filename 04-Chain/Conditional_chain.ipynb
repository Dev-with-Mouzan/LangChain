
# What is Conditional Chain in LangChain?

Conditional Chain is a way to execute a chain in LangChain based on a condition. It allows you to specify a condition and a chain to execute if the condition is true, and another chain to execute if the condition is false.

For example, you can use Conditional Chain to ask a user if they want to proceed with a certain action, and if they say yes, execute one chain, and if they say no, execute another chain.

Conditional Chain is useful when you need to handle different scenarios based on user input or other conditions. It allows you to create more complex and dynamic workflows in LangChain.


```python
from langchain_groq import ChatGroq
from langchain_core.prompts import PromptTemplate
from langchain_core.runnables import RunnableBranch
from langchain_core.output_parsers import StrOutputParser
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
model=ChatGroq(model="llama-3.1-8b-instant")
```

```python
summary_prompt=PromptTemplate(
    input_variables=["text"],
    template="Summarize the following text: {text}")
Explanation_prompt=PromptTemplate(
    input_variables=["text"],
    template="Explain the following text: {text}")

```

```python
parser=StrOutputParser()
```

```python
summary=summary_prompt | model | parser
explanation=Explanation_prompt | model | parser
```

```python
conditional_chain=RunnableBranch(
    (lambda text: len(text)>20, summary),
     explanation
)
```

```python
conditional_chain.get_graph().print_ascii()
```

```python
response_true_first_condition=conditional_chain.invoke({"text":"what is langchain?"})
```

```python
response_true_first_condition
```

```python
response_false_first_condition=conditional_chain.invoke({"text":"Langchain is a framework for building applications with language models. It provides a set of tools and abstractions that make it easier to work with language models and build complex applications."})
```

```python
response_false_first_condition
```
