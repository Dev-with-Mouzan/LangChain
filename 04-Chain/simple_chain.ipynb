# What is Simple Chain ?

A Simple Chain is a LangChain concept that allows you to chain together multiple LLMs to generate text based on human instructions. It is a powerful tool for generating code, writing articles, and creating chatbots. Simple Chains can be thought of as a series of function calls, where the output of one function call is used as the input for the next function call.


```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import PromptTemplate
from dotenv import load_dotenv
from langchain_core.output_parsers import StrOutputParser
```

```python
load_dotenv()
```

```python
model=ChatGoogleGenerativeAI(model="gemini-2.5-flash-lite")
```

```python
prompt=PromptTemplate(
    input_variables=["text"],
    template="Explain the following text:\n\n{text}")
```

```python
parser=StrOutputParser()
```

```python
simple_chain=prompt|model|parser
```

```python
simple_chain.get_graph().print_ascii()
```

```python
response=simple_chain.invoke({"text":"What is LangChain?"})
```

```python
print(response)
```
