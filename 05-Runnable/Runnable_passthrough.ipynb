
# What is Runnable Passthrough in LangChain

Runnable Passthrough is a way to encapsulate a unit of work that can be executed by the Langchain engine, and pass through additional information from the previous step to the next step.
in simple word output is same to input


```python
from langchain_core.runnables import RunnablePassthrough

def fake_llm(prompt: str) -> str:  # Fake LLM for the example
    return "completion"

runnable = {
    "llm1": fake_llm,
    "llm2": fake_llm,
} | RunnablePassthrough.assign(
    total_chars=lambda inputs: len(inputs["llm1"] + inputs["llm2"])
)

runnable.invoke("hello")
```
