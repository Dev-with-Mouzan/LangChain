
## What is Runnable Sequence?

A Runnable sequence is a list of Runnables that are executed in order. It is a way to define a complex workflow by combining multiple Runnables into a single unit of work. Runnable sequences are useful when you need to perform a series of tasks that depend on the output of the previous task.

For example, you can define a Runnable sequence that retrieves data from a database, processes the data, and then stores the results in a file. The first Runnable in the sequence would retrieve the data from the database, the second Runnable would process the data, and the third Runnable would store the results in a file.

By using Runnable sequences, you can create complex workflows that are easy to understand and maintain. You can also reuse individual Runnables in different sequences to create different workflows.


```python
from langchain_core.runnables import RunnableSequence
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
model=ChatGoogleGenerativeAI(model="gemini-2.5-flash-lite", temperature=0.7)
```

```python
prompt1=PromptTemplate(
    input_variables=["input"],
    template="What is the capital of {input}?"
)
prompt2=PromptTemplate(
    input_variables=["input"],
    template="What is the population of {input}?"
)
```

```python
parser=StrOutputParser()
```

```python
chain1=prompt1 | model | parser
chain2=prompt2 | model | parser
```

```python
sequence_runnable = RunnableSequence(chain1, chain2)
```

```python
response=sequence_runnable.invoke("France")
```
