# What is Langchain?
===================================================================



LangChain is an open-source orchestration framework designed to simplify the creation of applications using large language models (LLMs).

If an LLM like GPT-4 is the "brain," LangChain provides the nervous system, hands, and memory. It allows you to connect the model to your own data sources, give it the ability to interact with its environment (like searching the web), and "chain" together multiple steps to complete complex tasks.

# The Core Building Blocks
To build a solid GitHub repo, it helps to categorize LangChain into its primary components:

Model I/O: Standardizes the way you interact with various LLMs (OpenAI, Anthropic, Hugging Face). It handles Prompt Templates to manage inputs and Output Parsers to turn raw text into structured data (like JSON).

Retrieval: This is the foundation of RAG (Retrieval-Augmented Generation). It includes tools to load documents (PDFs, HTML), split them into chunks, and store them in Vector Databases for the model to "look up" facts it wasn't originally trained on.

Chains: The namesake of the library. Chains allow you to link multiple components together. While "Legacy Chains" were common, the modern standard is LCEL (LangChain Expression Language), which uses a Unix-like pipe syntax (|) to compose sequences.

Agents: Perhaps the most powerful part. Agents use an LLM as a "reasoning engine" to decide which tools to use (e.g., a calculator, a Google Search tool, or a Python REPL) to answer a specific query.

Memory: By default, LLMs are stateless (they don't remember previous messages). Memory components allow the chain to store and retrieve past interactions to maintain conversation context.

# Why use LangChain instead of raw API calls?
Modularity: You can swap out your LLM (e.g., moving from GPT-4 to a local Llama 3 model) by changing just one line of code.

Standardization: It provides a common interface for tasks that everyone needs, like document chunking or managing chat history.

Speed to Prototype: LangChain comes with "off-the-shelf" chains for common use cases like "Chat with your PDF" or "SQL Database Agent."
