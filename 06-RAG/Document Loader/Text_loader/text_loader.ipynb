
Text Loader
-------------

Text loader is a document loader that loads text files into the standardized document format. It is one of the simplest document loaders and is used to load text files.

The Text Loader takes a path to a text file and returns a document object containing the text and metadata. The metadata includes information such as the source, file name, and page number.

The Text Loader is useful for loading small text files, such as configuration files, or for loading text files that do not require any processing before being used.



```python
from langchain_community.document_loaders import TextLoader
```

```python
loader=TextLoader("sample_langchain_docs.txt")
```

```python
text = loader.load()
```

```python
text
```

```python
for text in loader.lazy_load(): # it is use to load the text file in lazy way 
    print(text)
```
