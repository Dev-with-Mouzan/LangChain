### What is PyPDFLoader
PyPDFLoader is a document loader that loads PDF files and returns a list of documents. It is used to load PDF files into a standardized document format which can then be split, embedded, and stored in vector databases.


```python
from langchain_community.document_loaders import PyPDFLoader
```

```python
loader=PyPDFLoader("MAD-6 -Basic Terms.pdf")
```

```python
data=loader.load()
```

```python
for i in data:
    print(i.page_content)
    print("----")
```
