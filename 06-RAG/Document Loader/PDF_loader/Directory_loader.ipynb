
# What is Directory loader

Directory loader is a document loader that loads files from a directory into a standardized document format. It takes a path to a directory as input and returns a list of documents. Each document contains the text content of the file and metadata such as the file name and path. The Directory loader is useful for loading files from a directory for semantic search or other applications.


```python
from langchain_community.document_loaders import DirectoryLoader, PyPDFLoader
```

```python
loader = DirectoryLoader(
    path="/",
    glob="*.pdf",
    loader_cls=PyPDFLoader
    # one more parameter is lazy load which load document on the demand of the user
)
```

```python
documents = loader.load()
```

```python
for doc in documents:
    print(doc.page_content)
    print("----")
```
