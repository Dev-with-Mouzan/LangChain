
What is a Document Loader?
==========================

A document loader is a utility that loads data from various sources into a standardized document format. This data can then be split, embedded, and stored in vector databases. Document loaders are used to load data from local files, remote APIs, or web pages.

Common document loaders include:

*   TextLoader for .txt files
*   PyPDFLoader for PDF documents
*   CSVLoader for CSV files
*   WebBaseLoader for web pages

Each document loaded by LangChain contains two main components:

1.  page_content: The actual text
2.  metadata: Information such as source, file name, or page number

Text files are the simplest starting point for understanding how document loaders work. Loading a text file with LangChain involves reading the file and storing the contents in a document object. The document object contains the text of the file and metadata such as the source, file name, and page number.
