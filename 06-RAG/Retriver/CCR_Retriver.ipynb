
## What is CCR in LangChain?

CCR stands for Chain Core Ranker which is a ranking model used in LangChain. It is responsible for ranking the documents retrieved from the database based on the relevance to the instruction. The CCR model is typically a small language model that is fine-tuned on the specific task of ranking documents. It takes into account the instruction, the document, and the context to generate a ranking score for each document. The documents are then ranked based on the score and the top-k documents are returned to the user.


```python
from langchain_community.vectorstores import FAISS
from langchain_ollama import ChatOllama,OllamaEmbeddings
from langchain_classic.retrievers import ContextualCompressionRetriever
from langchain_classic.retrievers.document_compressors import LLMChainExtractor
from langchain_core.documents import Document
```

```python
doc=[
    Document(page_content="The Grand Canyon is a steep-sided canyon carved by the Colorado River in Arizona, United States.photosentics is a process by which plants convert light energy into chemical energy."),
    Document(page_content="The Eiffel Tower is a wrought-iron lattice tower on the Champ de Mars in Paris, France.photosynthesis occurs in the chloroplasts of plant cells."),
    Document(page_content="The Great Wall of China is a series of fortifications that stretch across northern China.photosynthesis is essential for the survival of life on Earth."),
    Document(page_content="Machu Picchu is an ancient Incan city located in the Andes Mountains of Peru.photosynthesis involves the absorption of light by chlorophyll.")]
```

```python
emb=OllamaEmbeddings(model="nomic-embed-text:latest")
vectorstore=FAISS.from_documents(documents=doc,embedding=emb)
llm=ChatOllama(model="llama3.2:latest ")
```

```python
compresor=LLMChainExtractor.from_llm(llm)
```

```python
retriver=vectorstore.as_retriever(search_type="similarity",search_kwargs={"k":3})
```

```python
cc_retriver=ContextualCompressionRetriever(base_retriever=retriver, base_compressor=compresor)
```

```python
query="Explain photosynthesis in plants."
result=cc_retriver.invoke(query)
for doc in result:
    print(doc.page_content)
```
