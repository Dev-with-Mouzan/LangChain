
## What is MQR Retriver in Langchain?

The MQR Retriver is a type of Retriver in Langchain that uses the Maximum Query Relevance (MQR) algorithm to rank the retrieved documents based on their relevance to the instruction. The MQR algorithm is a type of relevance-based ranking algorithm that maximizes the relevance of the retrieved documents to the instruction.

The MQR Retriver is useful when the instruction is ambiguous or open-ended and requires the model to generate code based on multiple relevant code snippets. The MQR Retriver can retrieve a diverse set of code snippets that are relevant to the instruction, which can then be used by the generation model to generate code based on the instruction.

The MQR Retriver consists of three main components:

1. **Retrieval Model**: This is the model that is responsible for retrieving relevant code snippets from a database or knowledge base.
2. **Relevance Module**: This is the module that is responsible for computing the relevance of the retrieved code snippets to the instruction.
3. **Ranking Module**: This is the module that is responsible for ranking the retrieved code snippets based on their relevance to the instruction.


```python
from langchain_community.vectorstores import FAISS
from langchain_ollama import OllamaEmbeddings, ChatOllama
from langchain_classic.retrievers import MultiQueryRetriever
from langchain_core.documents import Document
```

```python
docs = [
    Document(page_content="Photosynthesis converts light into chemical energy."),
    Document(page_content="The Grand Canyon is a steep-sided canyon carved by the Colorado River."),
    Document(page_content="The Eiffel Tower is in Paris."),
]
```

```python
embeddings = OllamaEmbeddings(model="nomic-embed-text:latest")
vectorstore = FAISS.from_documents(docs, embeddings)
```

```python
mqr = MultiQueryRetriever.from_llm(
    retriever=vectorstore.as_retriever(),
    llm=ChatOllama(model="llama3.2:latest")
)
```

```python
query = "Explain photosynthesis and mention famous landmarks"
results = mqr.invoke({"query": query})

for doc in results:
    print(doc.page_content)
```
