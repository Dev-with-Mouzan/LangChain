# What is Wikipedia Retriver
Wikipedia Retriver is a web service by the Wikimedia Foundation that converts Markdown, content into HTML. It is used to generate HTML pages for the web,2 and is also used by the wikipedia software to generate HTML pages from Markdown content.


```python
import wikipedia
from langchain_community.retrievers import WikipediaRetriever
```

```python
retriver=WikipediaRetriever(top_k_result=3,lang="en")
```

```python
Query ="All books on Genrative Ai"
```

```python
docs=retriver.invoke(Query)
print(docs)

for i ,d in enumerate(docs): # Enumerate to get index and value (induc,value)
    print("Result # ",i+1)
    print("Content \n",d.page_content)
```
