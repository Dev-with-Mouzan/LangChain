
## What is MMR Retriver in Langchain?

The MMR Retriver is a type of Retriver in Langchain that uses the Maximum Marginal Relevance (MMR) algorithm to rank the retrieved documents based on their relevance to the instruction. The MMR algorithm is a type of diversity-based ranking algorithm that maximizes the marginal relevance of the retrieved documents.

The MMR Retriver is useful when the instruction is ambiguous or open-ended and requires the model to generate code based on multiple relevant code snippets. The MMR Retriver can retrieve a diverse set of code snippets that are relevant to the instruction, which can then be used by the generation model to generate code based on the instruction.

The MMR Retriver consists of three main components:

1. **Retrieval Model**: This is the model that is responsible for retrieving relevant code snippets from a database or knowledge base.
2. **Diversity Module**: This is the module that is responsible for computing the diversity of the retrieved code snippets.
3. **Ranking Module**: This is the module that is responsible for ranking the retrieved code snippets based on their relevance and diversity.

The MMR Retriver is a powerful tool for generating code based on human instructions. It can retrieve a diverse set of relevant code snippets and rank them based on their relevance and diversity, which can then be used by the generation model to generate code based on the instruction.


```python
from langchain_core.documents import Document
from langchain_community.vectorstores import FAISS
from langchain_ollama.embeddings import OllamaEmbeddings
```

```python
doc=[
    Document(page_content="Langchain i use to build applications "),
    Document(page_content="Langchain is used to build applications that are powered by language models. "),
    Document(page_content="Pakistan is my country "),
    Document(page_content="I love my country Pakistan "),
    Document(page_content="Artificial Intelligence is the future of technology. "),
    Document(page_content="Machine Learning is a subset of Artificial Intelligence. ")
    
    ]
```

```python
emb=OllamaEmbeddings(model="nomic-embed-text:latest")
```

```python
vector_store=FAISS.from_documents(documents=doc,embedding=emb)
```

```python
query="Which country do I love?"
```

```python
retriver=vector_store.as_retriever(search_type="mmr",search_kwargs={"k":2,"lambda_mul":0.5})
```

```python
result=retriver.invoke(query)
for doc in result:
    print(doc.page_content)
```

The Results are different from each other due to MMR
