## What are Retriver in LangChain?

A Retriver in LangChain is a component that is responsible for retrieving relevant information from a database or knowledge base. It is typically a small language model that is responsible for generating code based on human instructions.

The Retriver is the foundation of RAG (Retrieval-Augmented Generation). It includes tools to load documents (PDFs, HTML), split them into chunks, and store them in Vector Databases for the model to "look up" facts it wasn't originally trained on.

Retrieval models are used to generate code based on human instructions. They are typically small language models that are responsible for generating code based on human instructions.

The three components of the Retriver architecture interact with each other to generate code based on human instructions. The retrieval model is responsible for retrieving relevant code snippets from a database or knowledge base. The generation model is then responsible for generating code based on the retrieved code snippets and human instructions. Finally, the post-processing module is responsible for processing the output of the generation model to generate clean and correct code.


### Difference between LLM's & Retriver
The main difference between an LLM and a Retriver is that an LLM is responsible for generating code based on human instructions, while a Retriver is responsible for retrieving relevant information from a database or knowledge base.

An LLM is typically a large language model that is responsible for generating code based on human instructions. It is responsible for generating code that is both correct and relevant to the user's instruction.

On the other hand, a Retriver is a small language model that is responsible for retrieving relevant code snippets from a database or knowledge base. It is responsible for generating code that is relevant to the user's instruction, but may not necessarily be correct.

In RAG, the Retriver is responsible for retrieving relevant code snippets from a database or knowledge base, while the LLM is responsible for generating code based on the retrieved code snippets and human instructions.
