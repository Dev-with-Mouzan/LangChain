
## What is Vector Database Retriver in Langchain?

A Vector Database Retriver is a type of Retriver in Langchain that uses a vector database to store and retrieve dense vectors. The vector database is typically a FAISS database, which is a library for efficient similarity search and clustering of dense vectors.

The Vector Database Retriver is useful when the instruction is open-ended and requires the model to generate code based on multiple relevant code snippets. The Vector Database Retriver can retrieve a diverse set of code snippets that are relevant to the instruction, which can then be used by the generation model to generate code based on the instruction.

```python
from langchain_community.vectorstores import FAISS
from langchain_ollama import OllamaEmbeddings
from langchain_core.documents import Document
```

```python
documents=[Document(page_content="LangChain help developer to build LLm "),
           Document(page_content="FAISS is a vector data base for llm "),
           Document(page_content="Embedding convert vector in to the high diemenion vestor ")]
```

```python
embeddings=OllamaEmbeddings(model="nomic-embed-text:latest")
vectorstore=FAISS.from_documents(documents=documents,embedding=embeddings)
print("Vector store created",vectorstore)
```

```python

retriever = vectorstore.as_retriever(search_kwargs={"k": 2})# here forcing retriver to return 2 document
q="What are component of langchain? "
result= retriever.invoke(q)
print(result)
```
