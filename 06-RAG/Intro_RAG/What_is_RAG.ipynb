
# What is RAG

RAG stands for Retrieval-Augmented Generation. It is an AI technology that combines the strengths of two AI models: a retrieval model and a generation model.

The retrieval model is responsible for fetching relevant information from a database or knowledge base. The generation model is responsible for generating text based on the information retrieved by the retrieval model.

RAG is used in many applications such as chatbots, virtual assistants, and language translation systems. It is a powerful technology that can help machines generate more accurate and informative text.



### Problem With Tradition Chatbots
The traditional chatbot before RAG was based on a rule-based system. It had a set of pre-defined rules and a database of responses. When a user asked a question, the chatbot would retrieve the relevant information from the database and generate a response based on the rules.

The problem with the traditional chatbot was that it was unable to generate responses that were not pre-defined in the database. This limited the ability of the chatbot to generate responses that were relevant and informative.

Another problem with the traditional chatbot was that it was unable to learn from user interactions. It was unable to adapt to new information and improve its responses over time.

RAG was developed to address these limitations of traditional chatbots. It uses a combination of natural language processing and machine learning to generate responses that are relevant and informative.



### Components of RAG

RAG consists of four main components: document loader, text splitter, vector store, and retriever.

#### Document Loader
The document loader is responsible for loading data from various sources into a standardized document format. Common document loaders include TextLoader for .txt files, PyPDFLoader for PDF documents, CSVLoader for CSV files, and WebBaseLoader for web pages.

#### Text Splitter
The text splitter is responsible for splitting large documents into smaller chunks before processing. This is done to improve retrieval accuracy, reduce token usage, and improve LLM performance. Common text splitters include CharacterTextSplitter, RecursiveCharacterTextSplitter, and TokenTextSplitter.

#### Vector Store
The vector store is responsible for storing document embeddings in a vector database such as FAISS, Chroma, or Pinecone. These embeddings are used for semantic search over documents.

#### Retriever
The retriever is responsible for fetching relevant information from a database or knowledge base based on user queries. It uses the document embeddings stored in the vector store to retrieve the most relevant documents.
