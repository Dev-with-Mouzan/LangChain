
## Why we use Rad

Rad is a powerful tool for building applications with large language models (LLMs) and is the primary library used for building RAG chains. It provides a simple and standardized interface for interacting with LLMs, making it easy to build complex applications that integrate multiple models. RAG also provides a set of pre-built components and tools for building common tasks, such as document chunking, vector databases, and chatbots. This makes it easier to focus on building the specific application logic, rather than re-inventing common components.

Additionally, RAG is open-source, which means that the community can contribute to its development and provide feedback on its design and implementation. This has led to a powerful and flexible library that is well-suited for building a wide range of AI applications.



## Example Use Case

* Intelligent Chatbots & Customer Support: RAG enables AI assistants to access internal product documentation, FAQs, and real-time order data to provide accurate, context-aware answers, as used in enterprise support systems.

* Enterprise Search & Knowledge Management: Instead of just searching for documents, RAG-powered systems like Glean can read and synthesize information from internal wikis, wikis, and policy documents to answer complex employee 
questions.

* Domain-Specific Research (Legal, Medical, Finance): RAG is crucial for accessing vast, specialized, and up-to-date repositories, such as medical literature for diagnosis support or case law for legal research.

* Data Analysis & Reporting: RAG can query SQL databases, analyze data, and summarize insights via natural language interactions.
* Fact Verification & Summarization: RAG checks AI-generated text against reliable, retrieved sources to ensure    accuracy and generates summaries of large, complex documents.

* Personalization & E-commerce: Companies use RAG to create tailored recommendations by analyzing user behavior against current inventory and product details.
