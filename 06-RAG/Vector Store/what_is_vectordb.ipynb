
## What are Vector Databases?

Vector databases are a type of NoSQL database that are optimized for storing and querying large collections of dense vectors. They are commonly used in applications such as image and video search, recommender systems, and natural language processing.

Vector databases are designed to efficiently store and query dense vectors, such as those generated by deep learning models. They are typically optimized for fast query performance and can handle large volumes of data.

Some common use cases for vector databases include:

* Image and video search
* Recommender systems
* Natural language processing
* Chatbots and virtual assistants

Vector databases are typically used in conjunction with other data storage systems, such as relational databases or key-value stores. They are often used to store and query dense vectors that are generated by machine learning models.

Some popular vector database systems include:

* FAISS
* Chroma
* Pinecone
* Qdrant
* Annoy



## Difference between structured database and vector database

Structured databases are designed to store data in a structured format, such as tables with defined schemas. They are typically optimized for fast query performance using structured query languages (SQL) and are well suited for applications that require ad-hoc querying and aggregation of data.

Vector databases, on the other hand, are designed to store and query dense vectors. They are typically optimized for fast query performance using similarity-based queries and are well suited for applications that require efficient similarity-based search.

The main differences between structured databases and vector databases are:

* Data format: Structured databases store data in a structured format, such as tables with defined schemas. Vector databases store data as dense vectors.
* Query language: Structured databases use structured query languages (SQL) to query data. Vector databases use similarity-based queries, such as cosine similarity or L2 distance.
* Use case: Structured databases are well suited for applications that require ad-hoc querying and aggregation of data. Vector databases are well suited for applications that require efficient similarity-based search.

Examples of structured databases include relational databases, such as MySQL, and NoSQL databases, such as MongoDB. Examples of vector databases include FAISS, Chroma, and Pinecone.
