
### What is Sentiment Base Text Splitter?

Sentiment Base Text Splitter is a type of text splitter that splits text based on sentiment. It is useful when you want to perform sentiment analysis on a document, and you want to split the text into positive and negative sentiment chunks.

For example, if you have a document with the following text:

"I love LangChain. It is a powerful tool for building language models. However, it can be difficult to use."

A sentiment base text splitter would split the text into three chunks:

Chunk 1: "I love LangChain"
Chunk 2: "It is a powerful tool for building language models"
Chunk 3: "However, it can be difficult to use"

The sentiment base text splitter would then classify each chunk as either positive or negative based on the sentiment expressed in the text.


```python
from langchain_experimental.text_splitter import SemanticChunker
from langchain_ollama import OllamaEmbeddings
Embedding=OllamaEmbeddings(model="nomic-embed-text:latest")
```

```python

text = """
LangChain helps developers build LLM applications.
It provides tools for chaining prompts and models.

FAISS is a vector database.
It is used for similarity search in high-dimensional spaces.

Pakistan's capital is Islamabad.
It is a planned city.
"""

```

```python
splitter = SemanticChunker(Embedding,breakpoint_threshold_type="percentile")
```

```python
chunks = splitter.create_documents([text])
```

```python
for chunk in chunks:
    print(f"----Chunk#----")
    print(chunk.page_content)
```
