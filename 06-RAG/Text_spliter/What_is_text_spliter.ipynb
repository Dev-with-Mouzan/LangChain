
# What is Text Splitter?

Text splitters are an essential component of LangChain's document processing pipeline. They take a document as input and split it into smaller units of text, such as sentences, paragraphs, or even individual words. This allows for more efficient processing and better results when performing semantic search or other types of information retrieval.

Text splitters are useful when you want to:

* Break down large documents into smaller, more manageable pieces
* Extract specific information or entities from text
* Perform operations on individual words or phrases, such as sentiment analysis or topic modeling

LangChain provides a range of text splitters, including:

* CharacterTextSplitter: Splits text based on individual characters
* RecursiveCharacterTextSplitter: Splits text based on individual characters, but also allows for recursive splitting
* TokenTextSplitter: Splits text based on tokens, such as words or phrases

By using text splitters, you can unlock the full potential of LangChain's document processing capabilities and build more powerful and efficient applications.
