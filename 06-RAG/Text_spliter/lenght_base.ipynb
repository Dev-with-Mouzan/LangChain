
### What is length base text splitting?

Length base text splitting is a technique used to split text into chunks of a fixed length. It is commonly used in natural language processing tasks such as text classification, sentiment analysis, and language generation.

The main advantage of length base text splitting is that it is simple and efficient. It does not require any additional information such as part-of-speech tags or named entities, and it can be computed in O(n) time complexity.

The main disadvantage of length base text splitting is that it is not suitable for tasks that require a deeper understanding of the text such as question answering or text summarization. This is because it does not take into account the meaning of the text, and it may split the text in a way that is not meaningful.

Examples of length base text splitting include:

* Splitting a text into chunks of 50 characters each
* Splitting a text into chunks of 100 words each
* Splitting a text into chunks of 200 tokens each


```python
from langchain_text_splitters import CharacterTextSplitter
```

```python
text="LangChain is a Python framework for large language models (LLMs) and other large language model (LLM) APIs. It is an open source framework for building and deploying LLM apps."
```

```python
spliter=CharacterTextSplitter(chunk_size=5,chunk_overlap=0,separator=" ")
```

```python
chunks=spliter.split_text(text)
```
