
# What is text structure
Text structure is the organization of text into a hierarchical structure of headings, paragraphs, lists, and other elements. It is used to make the content of a document more readable and easier to understand. Text structure is important for search engines, as it allows them to better understand the content of a document and provide more accurate search results. It is also important for accessibility, as it allows people with disabilities to better understand the content of a document.


```python

from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
spliter=RecursiveCharacterTextSplitter(chunk_size=50,chunk_overlap=0,separators=["\n\n","\n","."," ",""])
```

```python
text="LangChain is a powerful framework for building applications that can understand and generate human language. It provides a simple and intuitive interface for working with natural language processing (NLP) tasks, such as text classification, sentiment analysis, and language generation. With LangChain, you can easily create chatbots, virtual assistants, and other conversational AI applications that can interact with users in a natural and engaging way."
chunks=spliter.split_text(text)
print(len(chunks))
for i in chunks:
    print(i)
    print("--------------")
```
