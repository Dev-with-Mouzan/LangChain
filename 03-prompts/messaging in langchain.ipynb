# Messages in LangChain

Messages are the core building blocks for communicating with chat models in LangChain. They represent individual units of conversation with a specific role and content.

## Message Types in LangChain

### 1. HumanMessage

Represents a message from the human/user. Used for user inputs.

```python
from langchain.messages import HumanMessage

human_msg = HumanMessage(content="What is the capital of France?")
print(human_msg)
print(f"Type: {type(human_msg)}")
```

### 2. AIMessage

Represents a response from the AI/LLM. Can include additional metadata like tool calls.

```python
from langchain.messages import AIMessage

ai_msg = AIMessage(content="The capital of France is Paris.")
print(ai_msg)
print(f"Type: {type(ai_msg)}")
```

### 3. SystemMessage

Provides instructions or context to the AI. Sets the behavior/persona of the AI.

```python
from langchain.messages import SystemMessage

system_msg = SystemMessage(content="You are a helpful Python programming assistant.")
print(system_msg)
```

### 4. ToolMessage

Represents the result of a tool/function call.

```python
from langchain.messages import ToolMessage

tool_msg = ToolMessage(
    content="[{"temperature": "25", "condition": "sunny"}]",
    tool_call_id="tool_123"
)
print(tool_msg)
```

### 5. AIMessageChunk

Used for streaming responses - represents a piece of an AIMessage.

```python
from langchain.messages import AIMessageChunk

chunk = AIMessageChunk(content="Hello")
print(chunk)
```

## Using Messages with Chat Models

```python
from langchain_openai import ChatOpenAI
from langchain_core.messages import HumanMessage, SystemMessage

llm = ChatOpenAI(model="gpt-4")

messages = [
    SystemMessage(content="You are a helpful assistant."),
    HumanMessage(content="Hello!")
]

response = llm.invoke(messages)
print(f"Response type: {type(response)}")
print(f"Content: {response.content}")
```

## Message History and Conversation

Maintain conversation history for multi-turn chats.

```python
messages = [
    SystemMessage(content="You are a helpful coding assistant."),
    HumanMessage(content="What is a list in Python?"),
    AIMessage(content="A list in Python is an ordered, mutable collection of items."),
    HumanMessage(content="How do I create one?")
]

response = llm.invoke(messages)
print(response.content)
```

## Message Utilities

### Convert Messages to Dict

```python
from langchain.messages import HumanMessage, SystemMessage

msg = HumanMessage(content="Hello world")
msg_dict = msg.to_dict()
print(msg_dict)
```

### Create Messages from Dict

```python
from langchain.messages import HumanMessage

msg_dict = {"type": "human", "data": {"content": "Hello"}}
msg = HumanMessage.from_dict(msg_dict)
print(msg.content)
```

## Summary

| Message Type | Role | Use Case |
|-------------|------|----------|
| HumanMessage | User | User inputs/queries |
| AIMessage | AI | Model responses |
| SystemMessage | System | Instructions/persona |
| ToolMessage | Tool | Tool execution results |
| AIMessageChunk | AI | Streaming response chunks |
