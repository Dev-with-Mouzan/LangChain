### 1. Prompt Templates

Prompt templates are pre-defined recipes for generating prompts. They allow you to create reusable prompt structures with placeholders for dynamic values.

```python
from langchain_core.prompts import PromptTemplate
from langchain_google_genai import ChatGoogleGenerativeAI
from dotenv import load_dotenv


```

```python
load_dotenv()
```

```python
chat = ChatGoogleGenerativeAI(model="gemini-2.5-flash-lite")
template = PromptTemplate(
    input_variables=["topic"],
      template="Explain the topic {topic} in simple terms.don't use any technical jargon.your answer should be concise and easy to understand and make sure to provide examples to illustrate your explanation.")
```

```python
templated_prompt = template.invoke(Topic:=input("Enter a topic to explain: ")) 
```

```python
response = chat.invoke(templated_prompt)
print("AI Explanation:", response.content)
```
