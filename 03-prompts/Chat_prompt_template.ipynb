###  Chat Prompts

Chat prompts are used with chat models and consist of a series of messages with roles (system, user, assistant).

### Message in LangChain
There are three type of Messages in Langchain
1. HumanMessage
2. AIMessage
3. SystemMessage

Each message type is used to represent different participants in a conversation with a language model.


```python
from langchain_core.prompts import ChatPromptTemplate

chat_template = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful assistant."),
    ("human", "{user_input}")
])

messages = chat_template.format_messages(user_input="What is Python?")
print(messages)
```
