# What are Prompts in LangChain?

Prompts in LangChain are instructions or queries sent to a Large Language Model (LLM) to generate a desired response. They serve as the interface between human intent and AI output.

## Why Use Prompt Templates?

- **Reusability**: Create once, use many times
- **Consistency**: Ensure uniform prompt structure
- **Dynamic Values**: Easily insert variables
- **Separation of Concerns**: Keep prompts separate from application logic
